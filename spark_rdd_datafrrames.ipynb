RDD (Resilient Distributed Dataset) is a fundamental data structure in Apache Spark. It is an immutable distributed collection of objects that can be processed in parallel across a cluster of machines. RDDs provide fault-tolerance and are designed to handle large-scale data processing tasks.

Here are some key characteristics and operations related to RDDs in Spark:

Characteristics:
1. Distributed: RDDs are distributed across multiple nodes in a cluster, allowing for parallel processing of data.
2. Immutable: RDDs are read-only and cannot be modified. However, transformations can be applied to create new RDDs.
3. Resilient: RDDs are fault-tolerant and can recover from failures. They achieve this through lineage, which keeps track of the operations performed on the RDD.

Operations:
1. Transformations: RDDs support two types of transformations:
   - Narrow transformations: These transformations do not require data shuffling across partitions, such as map, filter, and union.
   - Wide transformations: These transformations require data shuffling across partitions, such as reduceByKey and groupByKey.

2. Actions: Actions trigger the execution of transformations and return a result or write data to an external storage system. Examples of actions include count, collect, and save.

3. Persistence: RDDs can be persisted in memory or stored on disk to optimize performance. This helps avoid re-computation of RDDs in case of iterative algorithms or when RDDs are used multiple times.

4. Caching: RDDs can be cached in memory to speed up repeated access. This is particularly useful when iterative algorithms or interactive data exploration is performed.

RDDs form the foundation of Spark's programming model, allowing for efficient and parallel processing of data. While RDDs are a powerful abstraction, Spark also provides higher-level APIs like DataFrames and Datasets, which offer optimized query execution and a more structured approach to data processing.

```python
rdd = sc.parallelize([('A', 1), ('B', 2), ('A', 3), ('B', 4)])
grouped_rdd = rdd.groupByKey()
for key, values in grouped_rdd.collect():
    print(key, list(values))
```

The use of groupByKey() and reduceByKey() in PySpark RDDs. The groupByKey() operation groups values by key, while the reduceByKey() operation performs a reduction on the values based on their keys.

Remember that reduceByKey() is more efficient than groupByKey() for large datasets, as it performs local aggregation on each partition before shuffling the data across the network.


```python
reduced_rdd = rdd.reduceByKey(lambda x, y: x + y)
for key, value in reduced_rdd.collect():
    print(key, value)
```
