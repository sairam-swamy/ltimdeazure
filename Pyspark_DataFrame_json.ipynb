```python
from pyspark.sql.session import SparkSession
spark = SparkSession.builder.getOrCreate()
```

```python
schema="""customer_id string,
    Gender string,
    Name string,
    City string,
    `State Code` string,
    State string,
    `Zip Code` string,
    Country string,
    Continent string,
    Birthday date,
    year     string,
    month    string,
    week_name string
"""
```

```python
customers_json_df=spark.read.json(path='customers.json',schema=schema)
```

```python
customers_new_df=customers_json_df.withColumnsRenamed({'State Code':'State_Code','Zip Code':'zip_code'})
```

```python
customers_new_df.show()
```

```python
orders_df=spark.read.format('csv').option('header',True).load(path='orders.csv')
```

```python
orders_df.show()
```

```python
orders_df.groupBy('ProductKey').count().orderBy('count',ascending=False).show()
```

```python
orders_df.groupBy('ProductKey').agg({'ProductKey':'count','Quantity':'max'}).show()
```

```python
orders_df.groupBy('ProductKey').\
  agg({'ProductKey':'count','Quantity':'max'}).\
  withColumnRenamed('count(ProductKey)','count').\
  withColumnRenamed('max(Quantity)','max_quantity').\
  orderBy(['count','max_quantity'],ascending=False).show()
```

```python
orders_df.groupBy('ProductKey').\
  agg({'ProductKey':'count','Quantity':'max'}).\
  withColumnRenamed('count(ProductKey)','count').\
  withColumnRenamed('max(Quantity)','max_quantity').\
  orderBy(['count','max_quantity'],ascending=False).show()
```

```python
customers_new_df.\
  join(orders_df,on=customers_new_df.customer_id==orders_df.CustomerKey).\
  show()
```

```python
c_df=customers_new_df.alias('c_df')
O_df=orders_df.alias('O_df')
join_df=c_df.join(O_df,on=c_df.customer_id==O_df.CustomerKey)
```

```python
join_df.show()
```

```python
join_df.groupBy('c_df.country').count().orderBy('count',ascending=False).show()
```

```python
products_df=spark.read.format('csv').option('header',True).load(path='Products.csv',inferSchema=True)
```

```python
products_df.dtypes
```

```python
products_df.show()
```

```python
products_df.select('Unit Price USD',products_df['Unit Price USD'].cast('float')).show()
```

```python
from pyspark.sql.functions import regexp_replace
```

```python
products_df.select('Unit Price USD',regexp_replace(products_df['Unit Price USD'],r'\$','').cast('float')).show()
```

```python
products_df_new=products_df.\
                  withColumn('Unit Price USD',regexp_replace(products_df['Unit Price USD'],r'\$','').cast('float'))
```

```python
products_df_new.show()
```

```python
joined_df_product=join_df.join(products_df_new,on=join_df.ProductKey==products_df_new.ProductKey)
```

```python
joined_df_product.show()
```

```python
joined_df_updated=joined_df_product.withColumn('total_price',joined_df_product['Unit Price USD']*joined_df_product['Quantity'])
```

```python
from pyspark.sql.functions import col,count,sum
```

```python
joined_df_updated.groupBy('c_df.country').\
  agg(count('*'),sum('total_price')).\
  orderBy('count(1)',ascending=False).\
  show()
```

```python
joined_df_updated.groupBy('c_df.country').\
  agg(count('*'),sum('total_price')).\
  withColumn('sum_total_price',col('sum(total_price)').cast('long')).\
  orderBy('count(1)',ascending=False).\
  show()
```

```python
joined_df_updated.groupBy('c_df.country','state').\
  agg(count('*'),sum('total_price')).\
  withColumn('sum_total_price',col('sum(total_price)').cast('long')).\
  orderBy(['country','count(1)'],ascending=False).\
  show()
```

```python
from pyspark.sql import Window
from pyspark.sql.functions import row_number,rank,dense_rank
```

```python
c_df.show()
```

```python
window1=Window.\
          partitionBy('Country','year').\
          orderBy('month')
```

```python
c_df.withColumn('rank',rank().over(window1)).select('rank','name','Country','year','month').show()
```

```python
c_df.\
  withColumn('rank',rank().over(window1)).\
  select('rank','name','Country','year','month').\
  where('country="Australia" and year=1935').\
  show()
```

```python
c_df.\
  withColumn('rank',rank().over(window1)).\
  select('rank','name','Country','year','month').\
  where('country="Australia" and year=1936').\
  show()
```

```python
c_df.\
  withColumn('month',c_df['month'].cast('int')).\
  withColumn('rank',rank().over(window1)).\
  select('rank','name','Country','year','month').\
  where('country="Australia" and year=1936').\
  show()
```

```python
c_df.\
  withColumn('month',c_df['month'].cast('int')).\
  withColumn('rank',rank().over(window1)).\
  withColumn('dense_rank',dense_rank().over(window1)).\
  select('dense_rank','rank','name','Country','year','month').\
  where('country="Australia" and year=1936').\
  show()
```

```python
c_df.\
  withColumn('month',c_df['month'].cast('int')).\
  withColumn('rank',rank().over(window1)).\
  withColumn('dense_rank',dense_rank().over(window1)).\
  withColumn('row_number',row_number().over(window1)).\
  select('row_number','dense_rank','rank','name','Country','year','month').\
  where('country="Australia" and year=1936').\
  show()
```
